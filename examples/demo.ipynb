```python
import numpy as np
import pandas as pd
import lazy_visual as lv

rng = np.random.default_rng(0)
n = 300
df = pd.DataFrame({
    "group": rng.choice(["a", "b", "c"], n),
    "score": rng.normal(0, 1, n) + rng.choice([0, 1, 2], n),
    "x": rng.normal(0, 1, n),
    "y": rng.normal(0, 1, n),
    "class": rng.choice([0, 1], n),
    "s1": rng.uniform(0, 1, n),
    "s2": rng.uniform(0, 1, n),
})
df.head()
```

# lazy-visual demo

Run all cells (`Runtime → Run all` on Colab).


```python
_ = lv.scatterplot(df, x="x", y="y", hue="group", fit="linear")
```

```python
_ = lv.distribution(df, x="score", hue="group", kind="both")
```

```python
_ = lv.box(df, y="score", x="group", pairs=[("a","b"),("b","c")])
```

```python
_ = lv.violin(df, y="score", x="group", pairs=[("a","b"),("b","c")])
```

```python
_ = lv.raincloud(df, y="score", x="group", pairs=[("a","b"),("b","c")])
```

```python
_ = lv.heatmap(df, corr="pearson")
```

```python
_ = lv.upset({"A": {1,2,3,4,5}, "B": {3,4,5,6}, "C": {4,5,6,7}})
```

```python
_ = lv.roc(df, y_true="class", y_score=["s1","s2"])
```
